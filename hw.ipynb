```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([0, 5, 10, 15, 20, 25]).reshape(-1, 1)
y = np.array([55.0, 46.5, 35.0, 26.5, 18.0, 12.0])
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
new_temp = np.array([[12]])
predicted_energy = model.predict(new_temp)

print(f"Прогнозоване споживання для {new_temp[0][0]}°C: {predicted_energy[0]:.2f} кВт·год")
```

```python
y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', s=50, label='Історичні дані (Фактичні виміри)')
plt.plot(X, y_pred, color='red', linewidth=2, label='Модель (Лінійна регресія)')
plt.scatter(new_temp, predicted_energy, color='green', s=150, marker='*', zorder=5, label='Прогноз (12°C)')

plt.title('Залежність споживання енергії від температури')
plt.xlabel('Температура на вулиці (°C)')
plt.ylabel('Споживання електроенергії (кВт·год)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
